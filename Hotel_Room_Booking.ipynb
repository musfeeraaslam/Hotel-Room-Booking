```python
def calculate_hotel_bill(room_type, nights, breakfast=False):
    if room_type.lower() == "standard":
        rate = 5000
    elif room_type.lower() == "deluxe":
        rate = 8000
    elif room_type.lower() == "suite":
        rate = 15000
    else:
        print(f"Error: Unknown room type '{room_type}'")
        return

    base_bill = rate * nights

    if nights >= 7:
        discount_rate = 0.15
    elif nights >= 3:
        discount_rate = 0.10
    else:
        discount_rate = 0.00

    discount = base_bill * discount_rate
    bill_after_discount = base_bill - discount

    breakfast_cost = 1000 * nights if breakfast else 0
    final_bill = bill_after_discount + breakfast_cost

    print(f"Room Type: {room_type}")
    print(f"Rate per Night: {rate:.0f}")
    print(f"Nights: {nights}")
    print(f"Base Bill: {base_bill:.0f}")
    print(f"Discount: {discount:.0f} ({discount_rate * 100:.0f}%)")
    print(f"Breakfast: {'Yes' if breakfast else 'No'}")
    if breakfast:
        print(f"Breakfast Cost: {breakfast_cost:.0f}")
    print(f"Final Bill: {final_bill:.0f}")


room_type = input("Enter room type (Standard/Deluxe/Suite): ")
nights    = int(input("Enter number of nights: "))
bf_input  = input("Add breakfast? (yes/no): ").strip().lower()
breakfast = bf_input == "yes"

calculate_hotel_bill(room_type, nights, breakfast)
```
